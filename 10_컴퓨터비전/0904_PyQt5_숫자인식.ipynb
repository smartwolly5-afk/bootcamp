```python
import numpy as np
import tensorflow as tf
import tensorflow.keras.datasets as ds

from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense
from tensorflow.keras.optimizers import Adam

(x_train,y_train),(x_test,y_test)=ds.mnist.load_data()
x_train=x_train.reshape(60000,784)
x_test=x_test.reshape(10000,784)
x_train=x_train.astype(np.float32)/255.0
x_test=x_test.astype(np.float32)/255.0
y_train=tf.keras.utils.to_categorical(y_train,10)
y_test=tf.keras.utils.to_categorical(y_test,10)

dmlp=Sequential()
dmlp.add(Dense(units=1024,activation='relu',input_shape=(784,)))
dmlp.add(Dense(units=512,activation='relu'))
dmlp.add(Dense(units=512,activation='relu'))
dmlp.add(Dense(units=10,activation='softmax'))

dmlp.compile(loss='categorical_crossentropy',optimizer=Adam(learning_rate=0.0001),metrics=['accuracy'])
hist=dmlp.fit(x_train,y_train,batch_size=128,epochs=50,validation_data=(x_test,y_test),verbose=2)
print('정확률=', dmlp.evaluate(x_test,y_test,verbose=0)[1]*100)

dmlp.save('dmlp_trained.h5')

import matplotlib.pyplot as plt

plt.plot(hist.history['accuracy'])
plt.plot(hist.history['val_accuracy'])
plt.title('Accuracy graph')
plt.xlabel('epochs')
plt.ylabel('accuracy')
plt.legend(['train','test'])
plt.grid()
plt.show()

plt.plot(hist.history['loss'])
plt.plot(hist.history['val_loss'])
plt.title('Loss graph')
plt.xlabel('epochs')
plt.ylabel('loss')
plt.legend(['train','test'])
plt.grid()
plt.show()
```

```python
import cv2 as cv
import matplotlib.pyplot as plt
import numpy as np
import tensorflow as tf
from tensorflow.keras import datasets as ds

try:
    import winsound
except ImportError:
    winsound = None

# 1. Fashion-MNIST 데이터셋 및 학습된 CNN 모델 로드
(x_train, y_train), (x_test, y_test) = ds.fashion_mnist.load_data()

# 저장하신 파일명 'dmlpf_trained.h5' 적용
model = tf.keras.models.load_model('./dmlpf_trained.h5')

# Fashion-MNIST 10개 클래스 라벨
class_labels = [
    'T-shirt', 'Trouser', 'Pullover', 'Dress', 'Coat',
    'Sandal', 'Shirt', 'Sneaker', 'Bag', 'Ankle boot'
]

# 10개 클래스별 대표 예시 이미지 미리 생성 (40x40 크기)
sample_images = []
for i in range(10):
    idx = np.where(y_train == i)[0][0]
    sample_img = 255 - x_train[idx]  # 반전
    sample_img = cv.resize(sample_img, (40, 40), interpolation=cv.INTER_AREA)
    sample_img = cv.cvtColor(sample_img, cv.COLOR_GRAY2BGR)
    sample_images.append(sample_img)

# GUI 레이아웃 설정 (박스 3개, 크기 180x180)
BOX_COUNT = 3
BOX_SIZE = 180
MARGIN = 25

def reset():
    global img
    # 캔버스 크기: 너비 650, 높이 520
    img = np.ones((520, 650, 3), dtype=np.uint8) * 255
    
    # 1. 상단: 그리기 박스 3개
    for i in range(BOX_COUNT):
        x1 = MARGIN + i * (BOX_SIZE + MARGIN)
        x2 = x1 + BOX_SIZE
        cv.rectangle(img, (x1, 60), (x2, 60 + BOX_SIZE), (0, 0, 255), 2)
        
    # 2. 상단: 단축키 안내 문구
    cv.putText(img, 'e:erase  s:show  r:recognition  q:quit', (25, 35),
               cv.FONT_HERSHEY_SIMPLEX, 0.8, (255, 0, 0), 2)
    
    # 3. 하단: 예시 이미지 영역 구분선
    cv.line(img, (15, 360), (635, 360), (200, 200, 200), 2)
    cv.putText(img, '[ Reference Examples (0~9) ]', (25, 385),
               cv.FONT_HERSHEY_SIMPLEX, 0.6, (100, 100, 100), 2)

    # 4. 하단: 10개 클래스 예시 이미지 및 이름 배치 (2줄 x 5개)
    for i in range(10):
        row = i // 5
        col = i % 5
        x_start = 25 + col * 120
        y_start = 400 + row * 55
        
        # 40x40 예시 이미지 삽입
        img[y_start:y_start+40, x_start:x_start+40] = sample_images[i]
        
        # 라벨 텍스트 표시
        label_text = f"{i}:{class_labels[i]}"
        cv.putText(img, label_text, (x_start + 45, y_start + 25),
                   cv.FONT_HERSHEY_SIMPLEX, 0.4, (50, 50, 50), 1)

def grab_numerals():
    numerals = []
    for i in range(BOX_COUNT):
        x1 = MARGIN + i * (BOX_SIZE + MARGIN)
        roi = img[61:60 + BOX_SIZE, x1 + 1:x1 + BOX_SIZE, 0]
        
        # 28x28 리사이징 및 색상 반전
        roi = 255 - cv.resize(roi, (28, 28), interpolation=cv.INTER_CUBIC)
        numerals.append(roi)
    return np.array(numerals)

def show():
    numerals = grab_numerals()
    plt.figure(figsize=(12, 4))
    for i in range(BOX_COUNT):
        plt.subplot(1, BOX_COUNT, i + 1)
        plt.imshow(numerals[i], cmap='gray')
        plt.xticks([]); plt.yticks([])
    plt.show()

def recognition():
    numerals = grab_numerals()
    
    # CNN 입력을 위한 (3, 28, 28, 1) 형태 변환 및 정규화
    inputs = numerals.reshape(BOX_COUNT, 28, 28, 1).astype(np.float32) / 255.0
    res = model.predict(inputs)

    for i in range(BOX_COUNT):
        # 확률 상위 2개 인덱스 추출
        top_indices = np.argsort(res[i])[::-1]
        
        first_idx = top_indices[0]
        first_prob = res[i][first_idx] * 100
        
        second_idx = top_indices[1]
        second_prob = res[i][second_idx] * 100

        x1 = MARGIN + i * (BOX_SIZE + MARGIN)
        
        # 1위 예측 (이름 + 확률%)
        text_top1 = f"{class_labels[first_idx]} ({first_prob:.1f}%)"
        cv.putText(img, text_top1, (x1 + 5, 275),
                   cv.FONT_HERSHEY_SIMPLEX, 0.65, (255, 0, 0), 2)
        
        # 2위 예측 (이름 + 확률%)
        text_top2 = f"2nd: {class_labels[second_idx]} ({second_prob:.1f}%)"
        cv.putText(img, text_top2, (x1 + 5, 305),
                   cv.FONT_HERSHEY_SIMPLEX, 0.5, (100, 100, 100), 1)

    if winsound:
        winsound.Beep(1000, 500)

BrushSiz = 7
LColor = (0, 0, 0)

def writing(event, x, y, flags, param):
    # 그리기 영역 제한 (상단 3개 입력 박스 내부만 가능)
    if 60 < y < 240:
        if event == cv.EVENT_LBUTTONDOWN:
            cv.circle(img, (x, y), BrushSiz, LColor, -1)
        elif event == cv.EVENT_MOUSEMOVE and flags == cv.EVENT_FLAG_LBUTTON:
            cv.circle(img, (x, y), BrushSiz, LColor, -1)

reset()
cv.namedWindow('Writing')
cv.setMouseCallback('Writing', writing)

while True:
    cv.imshow('Writing', img)
    key = cv.waitKey(1)
    if key == ord('e'):
        reset()
    elif key == ord('s'):
        show()
    elif key == ord('r'):
        recognition()
    elif key == ord('q'):
        break

cv.destroyAllWindows()
```

```python
import sys
import cv2 as cv
import numpy as np
import tensorflow as tf
from PyQt5.QtWidgets import *

# EfficientNet 전용 모듈 및 전처리 함수 로드
from tensorflow.keras.applications.efficientnet import (
    EfficientNetB0,
    decode_predictions,
    preprocess_input,
)

try:
    import winsound
except ImportError:
    winsound = None

# ImageNet 가중치가 적용된 EfficientNetB0 모델 로드
model = EfficientNetB0(weights='imagenet')


class VehicleRecognitionApp(QMainWindow):

    def __init__(self):
        super().__init__()
        self.setWindowTitle('AI 교통수단 & 탈것 인식기')
        self.setGeometry(200, 200, 700, 100)

        fileButton = QPushButton('탈것 사진 열기', self)
        recognitionButton = QPushButton('교통수단 인식', self)
        quitButton = QPushButton('나가기', self)

        fileButton.setGeometry(10, 10, 100, 30)
        recognitionButton.setGeometry(110, 10, 100, 30)
        quitButton.setGeometry(510, 10, 100, 30)

        fileButton.clicked.connect(self.pictureOpenFunction)
        recognitionButton.clicked.connect(self.recognitionFunction)
        quitButton.clicked.connect(self.quitFunction)

        self.img = None

    def pictureOpenFunction(self):
        fname = QFileDialog.getOpenFileName(self, '탈것 사진 선택', './')

        if not fname[0]:
            return

        # 한글 경로 지원 (np.fromfile + cv.imdecode)
        img_array = np.fromfile(fname[0], np.uint8)
        self.img = cv.imdecode(img_array, cv.IMREAD_COLOR)

        if self.img is None:
            QMessageBox.warning(
                self, '경고', '이미지 파일을 읽을 수 없습니다.'
            )
            return

        cv.imshow('Vehicle Image', self.img)

    def recognitionFunction(self):
        if self.img is None:
            QMessageBox.warning(
                self, '경고', '먼저 탈것/교통수단 사진을 열어주세요.'
            )
            return

        # 1. BGR -> RGB 변환 및 224x224 리사이즈
        img_rgb = cv.cvtColor(self.img, cv.COLOR_BGR2RGB)
        resized_img = cv.resize(img_rgb, (224, 224))

        # 2. 배치 차원 추가 및 EfficientNet 전처리
        x = np.expand_dims(resized_img, axis=0).astype(np.float32)
        x = preprocess_input(x)

        # 3. 모델 추론 및 상위 3개 예측 결과 추출
        preds = model.predict(x)
        top3 = decode_predictions(preds, top=3)[0]

        img_draw = self.img.copy()

        # 4. 결과 출력
        for i in range(3):
            class_name = top3[i][1]  # 예: 'minibus', 'sports_car' 등
            prob = top3[i][2] * 100  # 백분율(%)

            # 이름의 언더바(_)를 공백으로 변경하여 깔끔하게 표시
            clean_name = class_name.replace('_', ' ').title()
            text = f'{i+1}위: {clean_name} ({prob:.1f}%)'

            # 가독성을 위한 검은 테두리 + 흰 글씨
            cv.putText(
                img_draw,
                text,
                (15, 50 + i * 40),
                cv.FONT_HERSHEY_SIMPLEX,
                0.8,
                (0, 0, 0),
                4,
            )
            cv.putText(
                img_draw,
                text,
                (15, 50 + i * 40),
                cv.FONT_HERSHEY_SIMPLEX,
                0.8,
                (255, 255, 255),
                2,
            )

        cv.imshow('Vehicle Image', img_draw)
        if winsound:
            winsound.Beep(1000, 500)

    def quitFunction(self):
        cv.destroyAllWindows()
        self.close()


app = QApplication(sys.argv)
win = VehicleRecognitionApp()
win.show()
app.exec_()
```

```python
import sys
import cv2 as cv
import matplotlib.pyplot as plt
import numpy as np
import tensorflow as tf
from PyQt5.QtWidgets import *

# EfficientNet 전용 모듈 및 전처리 함수 로드
from tensorflow.keras.applications.efficientnet import (
    EfficientNetB0,
    decode_predictions,
    preprocess_input,
)

try:
    import winsound
except ImportError:
    winsound = None

# ImageNet 가중치가 적용된 EfficientNetB0 모델 로드
model = EfficientNetB0(weights='imagenet')


class VehicleRecognitionApp(QMainWindow):

    def __init__(self):
        super().__init__()
        self.setWindowTitle('AI 교통수단 & 탈것 인식기')
        self.setGeometry(200, 200, 700, 100)

        fileButton = QPushButton('탈것 사진 열기', self)
        recognitionButton = QPushButton('교통수단 인식', self)
        quitButton = QPushButton('나가기', self)

        fileButton.setGeometry(10, 10, 100, 30)
        recognitionButton.setGeometry(110, 10, 100, 30)
        quitButton.setGeometry(510, 10, 100, 30)

        fileButton.clicked.connect(self.pictureOpenFunction)
        recognitionButton.clicked.connect(self.recognitionFunction)
        quitButton.clicked.connect(self.quitFunction)

        self.img = None
        self.count = 0  # 인식 횟수 카운터 (행 위치 제어용)

        # Matplotlib 대화형 모드(plt.ion) 활성화
        plt.ion()
        self.fig = None

    def pictureOpenFunction(self):
        fname = QFileDialog.getOpenFileName(self, '탈것 사진 선택', './')

        if not fname[0]:
            return

        # 한글 경로 지원 (np.fromfile + cv.imdecode)
        img_array = np.fromfile(fname[0], np.uint8)
        self.img = cv.imdecode(img_array, cv.IMREAD_COLOR)

        if self.img is None:
            QMessageBox.warning(
                self, '경고', '이미지 파일을 읽을 수 없습니다.'
            )
            return

        QMessageBox.information(
            self,
            '안내',
            '이미지가 로드되었습니다. [교통수단 인식] 버튼을 눌러주세요.',
        )

    def recognitionFunction(self):
        if self.img is None:
            QMessageBox.warning(
                self, '경고', '먼저 탈것/교통수단 사진을 열어주세요.'
            )
            return

        # 1. BGR -> RGB 변환 및 224x224 리사이즈
        img_rgb = cv.cvtColor(self.img, cv.COLOR_BGR2RGB)
        resized_img = cv.resize(img_rgb, (224, 224))

        # 2. 배치 차원 추가 및 EfficientNet 전처리
        x = np.expand_dims(resized_img, axis=0).astype(np.float32)
        x = preprocess_input(x)

        # 3. 모델 추론 및 상위 3개 예측 결과 추출
        preds = model.predict(x)
        top3 = decode_predictions(preds, top=3)[0]

        # 데이터 정리 (상위 3개 클래스 이름 & 확률)
        class_names = [top3[i][1].replace('_', ' ').title() for i in range(3)]
        probs = [top3[i][2] * 100 for i in range(3)]

        # 4. Matplotlib 출력 제어 (누적 출력)
        self.count += 1

        # 첫번째 인식일 경우 Figure 생성
        if self.fig is None or not plt.fignum_exists(self.fig.number):
            self.fig = plt.figure(figsize=(10, 4 * self.count))
        else:
            # 기존 창 크기를 아래로 확장
            self.fig.set_size_inches(10, 4 * self.count)

        # 서브플롯 추가 (행: self.count 개, 열: 2 개)
        # 이미지 출력 (왼쪽)
        ax_img = self.fig.add_subplot(self.count, 2, (self.count - 1) * 2 + 1)
        ax_img.imshow(img_rgb)
        ax_img.set_title(f'Test #{self.count} Image')
        ax_img.axis('off')

        # 확률 막대그래프 출력 (오른쪽)
        ax_bar = self.fig.add_subplot(self.count, 2, (self.count - 1) * 2 + 2)
        bars = ax_bar.barh(class_names[::-1], probs[::-1], color='skyblue')
        ax_bar.set_xlim(0, 100)
        ax_bar.set_xlabel('Probability (%)')
        ax_bar.set_title(f'Top-3 Predictions (#{self.count})')

        # 수치 텍스트 표시
        for bar in bars:
            width = bar.get_width()
            ax_bar.text(
                width + 1,
                bar.get_y() + bar.get_height() / 2,
                f'{width:.1f}%',
                va='center',
            )

        plt.tight_layout()
        plt.draw()
        plt.pause(0.1)

        if winsound:
            winsound.Beep(1000, 500)

    def quitFunction(self):
        plt.close('all')
        cv.destroyAllWindows()
        self.close()


app = QApplication(sys.argv)
win = VehicleRecognitionApp()
win.show()
app.exec_()
```
