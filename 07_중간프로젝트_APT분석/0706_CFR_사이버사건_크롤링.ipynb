```python
import time
import random
from selenium import webdriver
from bs4 import BeautifulSoup
from selenium.webdriver.support.ui import Select

def get_cfr_glossary():
    # 전체화면용 옵션 설정
    options = webdriver.ChromeOptions()
    options.add_argument("--start-maximized")
    
    # 크롬 드라이버 실행
    wd = webdriver.Chrome(options=options)
    
    # 목적지 URL 설정 (CFR Cyber Operations Glossary)
    url = "https://www.cfr.org/cyber-operations/#Glossary"
    wd.get(url)
    
    # 페이지가 완전히 로딩될 때까지 임의의 시간 동안 대기 (2~5초)
    time.sleep(random.randint(2, 5))
    
    # BeautifulSoup을 이용해 현재 페이지의 HTML 소스 파싱
    html = wd.page_source
    get_t = BeautifulSoup(html, 'html.parser')
    
    # 2초 추가 대기 후 브라우저 객체 반환
    time.sleep(2)
    sponsor_dropdown = wd.find_element(By.CSS_SELECTOR, '[class*="field-operators"] select, [class*="state-sponsor"] select')
    select_sponsor = Select(sponsor_dropdown)

# 3. 'China' 선택하기 (2가지 방법 중 하나를 선택하세요)
# 방법 A: 화면에 보이는 텍스트로 선택 (추천)
    select_sponsor.select_by_visible_text("China")
    post_elements = get_t.select('.field-content.cst-issue.views-field-title a')
    post_list = []
    for post in post_elements:
        title = post.get_text(strip=True) # 제목 텍스트 (앞뒤 공백 제거)
        link = post.get('href')           # <a> 태그의 href 속성 값 (URL)
    
    # 딕셔너리 형태로 저장
        post_list.append({
            "title": title,
            "link": link
        })

    print(f"총 {len(post_list)}개의 게시물을 찾았습니다.\n")
    for idx, item in enumerate(post_list, 1):
        print(f"[{idx}] {item['title']}")
        print(f"    링크: {item['link']}")
        print("-" * 50)

    return wd
get_cfr_glossary()
# 함수 실행 예시
# driver = get_cfr_glossary()
```

```python
import time
import random
import csv
from selenium import webdriver
from selenium.webdriver.common.by import By
from selenium.webdriver.support.ui import Select
from bs4 import BeautifulSoup

def crawl_cfr_china_incidents():
    options = webdriver.ChromeOptions()
    options.add_argument("--start-maximized")
    
    # 1. 브라우저 실행 및 사이트 접속
    wd = webdriver.Chrome(options=options)
    url = "https://www.cfr.org/cyber-operations/"
    wd.get(url)
    
    time.sleep(random.randint(3, 5))
    
    # 2. 'China' 필터 적용하기
    print("▶ 'China' 필터 적용 중...")
    sponsor_dropdown = wd.find_element(By.CSS_SELECTOR, '[class*="field-operators"] select, [class*="state-sponsor"] select')
    select_sponsor = Select(sponsor_dropdown)
    select_sponsor.select_by_visible_text("China")
    
    # [중요] 필터 적용 후 결과 목록이 새로고침될 때까지 충분히 대기합니다.
    time.sleep(3) 
    
    # 3. HTML 파싱 및 링크 수집 (필터가 적용된 후의 HTML을 가져옵니다)
    print("▶ 게시물 링크 수집 중...")
    html = wd.page_source
    soup = BeautifulSoup(html, 'html.parser')
    
    post_elements = soup.select('.field-content.cst-issue.views-field-title a')
    post_list = []
    
    for post in post_elements:
        title = post.get_text(strip=True)
        link = post.get('href')
        # 상대 경로면 절대 경로로 변경
        if not link.startswith('http'):
            link = "https://www.cfr.org" + link
            
        post_list.append({"title": title, "link": link})

    print(f"총 {len(post_list)}개의 중국 관련 게시물을 찾았습니다.\n")
    
    # 4. 각 링크 방문하여 상세 내용 긁어오고 CSV 저장하기
    output_filename = "cfr_china_cyber_operations.csv"
    
    # CSV 파일 열기 (한글 깨짐 방지 utf-8-sig)
    with open(output_filename, mode='w', encoding='utf-8-sig', newline='') as f:
        writer = csv.writer(f)
        writer.writerow(["Title", "URL", "Description", "Details(Meta)"]) # 헤더 작성
        
        print("▶ 상세 본문 추출 및 CSV 저장을 시작합니다...")
        
        for idx, item in enumerate(post_list, 1):
            title = item['title']
            url = item['link']
            print(f"[{idx}/{len(post_list)}] 수집 중: {title}")
            
            try:
                # 상세 페이지로 이동
                wd.get(url)
                time.sleep(random.randint(2, 4)) # 차단 방지를 위한 랜덤 대기
                
                # 상세 페이지 파싱
                detail_html = wd.page_source
                detail_soup = BeautifulSoup(detail_html, 'html.parser')
                
                # 본문 추출 (p 태그들 긁어오기)
                p_tags = detail_soup.select('.article-body p, .field-name-body p, .content p')
                description = "\n".join([p.get_text(strip=True) for p in p_tags])
                
                if not description:
                    description = "본문 텍스트를 찾을 수 없습니다."
                
                # 메타 데이터(우측/하단 요약 정보) 추출
                meta_sections = detail_soup.select('.field, .row, [class*="field-name-field-"]')
                meta_details = []
                for meta in meta_sections:
                    meta_text = meta.get_text(": ", strip=True)
                    if meta_text and ":" in meta_text and len(meta_text) < 200:
                        meta_details.append(meta_text)
                meta_summary = " | ".join(list(set(meta_details)))
                
                # CSV에 한 줄 작성
                writer.writerow([title, url, description, meta_summary])
                
            except Exception as e:
                print(f"❌ '{title}' 크롤링 중 오류: {e}")
                writer.writerow([title, url, "오류 발생", str(e)])
                
    # 5. 작업 완료 후 브라우저 닫기
    wd.quit()
    print(f"\n🎉 크롤링 완료! 파일이 저장되었습니다: {output_filename}")

# 실행
crawl_cfr_china_incidents()
```

```python
import requests
from bs4 import BeautifulSoup
import pandas as pd
import re

def get_apt_groups(result):
    url = "https://en.wikipedia.org/wiki/Advanced_persistent_threat"
    headers = {
        'User-Agent': 'Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/120.0.0.0 Safari/537.36'
    }
    
    html = requests.get(url, headers=headers)
    
    if html.status_code != 200:
        print(f"웹페이지 접근 차단됨 (상태 코드: {html.status_code})")
        return
        
    soup = BeautifulSoup(html.text, 'html.parser')
    
    # [변경 항목] 복잡한 형제 노드 추적 대신, 페이지 내의 모든 h3와 ul을 먼저 확보합니다.
    # 위키피디아 본문 전체를 감싸는 최외각 div를 기준으로 잡습니다.
    content_div = soup.find('div', class_='mw-content-ltr') or soup.find(id='bodyContent')
    
    if not content_div:
        print("본문 구역을 찾을 수 없습니다.")
        return

    current_country = ""
    
    # 본문 안의 요소들을 순서대로 내려가며 탐색합니다.
    for element in content_div.find_all(['h3', 'div', 'ul'], recursive=True):
        
        # 1. 국가명 찾기 (h3 태그이거나, mw-heading3 클래스를 가진 div인 경우)
        is_h3 = element.name == 'h3'
        is_div_h3 = element.name == 'div' and 'mw-heading3' in ''.join(element.get('class', []))
        
        if is_h3 or is_div_h3:
            # 국가 이름만 깨끗하게 추출
            text = element.get_text(strip=True).replace('[edit]', '')
            # APT groups 섹션 아래에 있는 국가들만 타겟팅하기 위해 필터링 (예: China, Russia, Iran 등)
            # 만약 Mitigation 같은 다른 대섹션의 h3가 나오면 패스하기 위함
            if text and not any(word in text for word in ['Contents', 'See also', 'References']):
                current_country = text
                print(f"[{current_country}] 구역 발견, 데이터 추출 중...")
                
        # 2. 국가가 지정된 상태에서 ul(리스트)을 만나면 내부 li(그룹명) 추출
        elif element.name == 'ul' and current_country:
            # 이 'ul'이 APT 그룹 목록인지 검증하기 위해 내부 첫 번째 li 확인
            first_li = element.find('li')
            if first_li:
                # 위키피디아 특성상 다른 섹션의 ul과 섞이는 걸 방지 (일반적으로 APT 그룹 리스트는 굵은 글씨(b, strong)나 링크(a)로 시작함)
                for li in element.find_all('li', recursive=False):
                    raw_text = li.get_text(strip=False).split('\n')[0]
                    group_name = re.sub(r'\[.*?\]', '', raw_text).strip()
                    
                    if group_name and len(group_name) < 100: # 너무 긴 설명글 제외
                        result.append([current_country, group_name])

def main():
    result = []
    print("APT 그룹 데이터 수집을 시작합니다...\n")
    
    get_apt_groups(result)
    
    if result:
        df = pd.DataFrame(result, columns=['Country', 'APT_Group'])
        filename = 'APT_Groups_Static.csv'
        df.to_csv(filename, encoding='utf-8-sig', index=False)
        print(f'\n{filename} 데이터 저장 완료! 총 {len(df)}개 수집됨.')
        print(df.head(15)) # 15개 출력해서 확인
    else:
        print("수집된 데이터가 없습니다. 위키피디아의 태그 구조를 다시 확인해야 합니다.")

if __name__ == "__main__":
    main()
```
