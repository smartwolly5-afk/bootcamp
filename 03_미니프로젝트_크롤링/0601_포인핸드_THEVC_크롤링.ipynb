````python
from bs4 import BeautifulSoup
import urllib.request
import pandas as pd
import datetime
import requests
from selenium import webdriver
from selenium.webdriver.common.by import By
import time
from collections import defaultdict
import random
````

````python
my_id = 'YOUR_ID'
my_pw = 'YOUR_PASSWORD'
#전체화면용 옵션
options = webdriver.ChromeOptions()
#전체화면으로 키기
options.add_argument("--start-maximized")
wd = webdriver.Chrome(options=options)
wd.get("https://thevc.kr/")
time.sleep(random.randint(2,5))
````

````python
#추천입양동물 (수정본)

from selenium import webdriver
from selenium.webdriver.common.by import By
from selenium.webdriver.support.ui import Select
import pandas as pd
import time

url = "https://pawinhand.kr/"

def regions():
    wd = webdriver.Chrome()
    wd.get(url)
    wd.refresh()
    time.sleep(3)

    result = []

    # 팝업창 제거 
    try:
        close_btn = wd.find_element(By.CSS_SELECTOR, ".popup-close")
        close_btn.click()
        time.sleep(1)
    except:
        pass

    # 메인페이지에서 보호소로 들어가기 
    shelter_btn = wd.find_element(By.PARTIAL_LINK_TEXT, "보호소")
    wd.execute_script("arguments[0].click();", shelter_btn)
    time.sleep(2)

    # 추천 입양 동물 클릭
    shelterFind_btn = wd.find_element(By.PARTIAL_LINK_TEXT, "추천 입양 동물")
    wd.execute_script("arguments[0].click();", shelterFind_btn)
    time.sleep(2)

#필터로 안되니까 지역을 일일이 눌러줘야됨 
#시작점이 어딘지 끝점이 어딘지 , 1번만 돌려주면 됨

    selects = wd.find_elements(By.CSS_SELECTOR, ".swiper-wriper")

    for local in range(len(selects)):
        selects = wd.find_elements(By.CSS_SELECTOR, ".swiper-slide")               
        print(local, selects[local].text)



regions()
````

````python
login_button = wd.find_element(By.CSS_SELECTOR, 'button.box-button.main.medium')
login_button.click()
````

````python
id_box = wd.find_element(By.ID,'email')
id_box.send_keys(my_id)
pw_box = wd.find_element(By.ID,'password')
pw_box.send_keys(my_pw)
log_button = wd.find_element(By.CSS_SELECTOR,'button.box-button.main.large.mt-40.w-full')
log_button.click()
````

````python
html = wd.page_source
get_t=BeautifulSoup(html,'html.parser')
````

````python
#더보기 있으면 클릭하기
def more_data():
    # more = wd.find_element(By.CSS_SELECTOR, 'button.text-button.small.gray.paddingless')
    # if more:
    #     wd.execute_script("arguments[0].click();", more)
    # '조회수 랭킹'이라는 텍스트를 가진 h2가 포함된 부모 div를 찾고, 그 안의 버튼을 클릭
    more_btn = wd.find_elements(By.CSS_SELECTOR, "div[data-v-ecce4443] > button.text-button.small.gray.paddingless:last-child")

    if len(more_btn) > 0:
        wd.execute_script("arguments[0].scrollIntoView({block: 'center'});", more_btn[0])
        time.sleep(1)
        wd.execute_script("arguments[0].click();", more_btn[0])
````

````python
more_data()
````

````python
#더보기 들어가서 총 유치금액, 총 투자금액 뽑아오기
def c_all():
    try:
        price_block = wd.find_elements(By.CSS_SELECTOR, ".summary-wrap")
        price_text = price_block[0].find_element(By.CSS_SELECTOR, "b").text
        total_price = price_block[1].find_element(By.CSS_SELECTOR, 'b').text
        print(f"총 유치 금액: {price_text}")
        print(f"총 투자 대상:{total_price}")
        return price_text
    except Exception as e:
        return None
````

````python
c_all()
````

````python
def c_data_all():
    all_table = wd.find_elements(By.CSS_SELECTOR, "div.table-wrap.is-full-width tr > th")
    head = []
    for span in all_table:
        head.append(span.text)
    print(head)
    return head
````

````python
c_data_all()
````

````python
def c_data():
    rows = wd.find_elements(By.CSS_SELECTOR, "div.table-wrap.is-full-width tbody > tr")
    extracted_data = []
    for row in rows:
        # 2. 각 행 안에서 개별 셀(td, th)을 찾습니다.
        cells = row.find_elements(By.CSS_SELECTOR, "td, th")
        
        # 3. 셀의 텍스트만 뽑아서 리스트로 만듭니다.
        row_data = [cell.text for cell in cells]
        
        # 이제 row_data는 ['제품명', '상태', '금액'] 처럼 분리된 리스트입니다.
        print(row_data) 
        extracted_data.append(row_data)
    return extracted_data
````

````python
c_data()
````

````python
def save_to_csv(headers, data, filename):
    # 리스트 데이터를 데이터프레임으로 변환
    df = pd.DataFrame(data, columns=headers)
    
    df.to_csv(f"{filename}.csv", index=False, encoding='utf-8-sig')
    print(f"{filename}.csv 저장 완료!")

head = c_data_all()
data = c_data()

if head and data:
    save_to_csv(head, data, "ranking_data")
````

````python
#다음 페이지로 넘기고 만약 팝업창이 뜨면 끄고 원래 페이지로 복귀
def next_page():
    button_next = wd.find_elements(By.CSS_SELECTOR, "div.pagination > button")
    wd.execute_script("arguments[0].click();", button_next[3])
    time.sleep(2)
    close_modal()
def close_modal(wd, head,data,name,i):
    try:
        close_btn = wd.find_element(By.CSS_SELECTOR, "button.icon-button.medium.paddingless.close-button")
        
        # 버튼이 클릭 가능한 상태일 때 클릭
        wd.execute_script("arguments[0].click();", close_btn)
        time.sleep(1) # 모달이 완전히 닫힐 때까지 잠시 대기
        save_to_csv(head,data,name[i])
        wd.get("https://thevc.kr/")

    except:
        pass
````

````python
def next_daym(i):
    element = wd.find_element(By.XPATH, f"//h2[contains(., '{"투자/M&A 랭킹"}')]")
    mn_button = element.find_element(By.CSS_SELECTOR, 'button.text-button')
    wd.execute_script("arguments[0].scrollIntoView({block: 'center'});", mn_button)
    time.sleep(1)
    wd.execute_script("arguments[0].click();", mn_button)
    time.sleep(1)
    buttons_m = element.find_elements(By.CSS_SELECTOR, '.list-20 button')
    buttons_m[i].click()
````

````python
#카테고리 위치
tabs = wd.find_elements(By.CSS_SELECTOR, 'div.tab-content')
lank_tabs_l = tabs[14:25]
current_index_l = 0
lank_tabs_m = tabs[26:28]
current_index_m = 0
# print(tabs)
# for i, tab in enumerate(tabs):
#     print(f"{i}번째 탭 이름: {tab.text}")
````

````python
#다음 카테고리 클릭하고 실행
def next_tabl():
    global current_index_l
    if current_index_l < len(lank_tabs_l):
        target_tab = lank_tabs_l[current_index_l]
        cat_name = target_tab.text
        wd.execute_script("arguments[0].click();", target_tab)
        time.sleep(2)
        html = wd.page_source
        get_t=BeautifulSoup(html,'html.parser')
        current_index_l += 1
        return get_t ,cat_name
    else:
        return None, None
    # # start_tab = wd.find_element(By.CSS_SELECTOR,'div.tab-content')
    # # start_tab.click()
    # time.sleep(3)
    
````

````python
#다음 카테고리 클릭하고 실행
def next_tabm():
    global current_index_m
    if current_index_m < len(lank_tabs_m):
        target_tab = lank_tabs_m[current_index_m]
        cat_name = target_tab.text
        wd.execute_script("arguments[0].click();", target_tab)
        time.sleep(2)
        html = wd.page_source
        get_t=BeautifulSoup(html,'html.parser')
        current_index_m += 1
        return get_t ,cat_name
    else:
        return None, None
    # # start_tab = wd.find_element(By.CSS_SELECTOR,'div.tab-content')
    # # start_tab.click()
    # time.sleep(3)
    
````

````python
#랭킹 별로 저장하기
all_data = []
for i in range(10):
    time.sleep(2)
    result = next_tabl()
    get_t, cat_name = result
    items = get_t.select('ol.list-wrap > li')
# print(items)
    now_tab = []
    num = 0
    for item in items:
        if num < 10:
            rank_d = {}
            link_tag = item.select_one('a')
            name = link_tag.select_one('div.text-truncate.text-16')
            company = name.contents[0].strip()
            product = name.select_one('div.text-truncate')
            # print(company)
            # print(product.text)
            num += 1
            now_tab.append({company:product.text})
    all_data.append({"tab_name":cat_name,
                     "rank": now_tab})    
print(all_data)   
````

````python
flat_data = []
for tab in all_data:
    tab_name = tab['tab_name']
    for rank_item in tab['rank']:
        # rank_item은 {'기업명': '제품명'} 형태입니다.
        for company, product in rank_item.items():
            flat_data.append({
                "탭이름": tab_name,
                "기업명": company,
                "제품명": product
            })

# 2. DataFrame으로 변환
df = pd.DataFrame(flat_data)

# 3. CSV로 저장 (인코딩을 cp949로 해야 엑셀에서 한글이 깨지지 않습니다)
df.to_csv(f"startup_rankings_{i}.csv", index=False, encoding='cp949')

print("CSV 저장 완료!")
````
