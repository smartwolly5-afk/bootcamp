```python
def factorial(n):
    output = 1
    for i in range(1, n+1):
        output *= i
    return output
print("2!:", factorial(2))
print("5!:", factorial(5))
```

```python
def factorial(n):
    if n == 0:
        return 1
    else:
        print(f"나 {n}번째 팩토리얼")
        return n * factorial(n - 1)

print("2!:", factorial(2))
print("3!:", factorial(3))
print("4!:", factorial(10))
```

```python
counter = 0
dictionary = {
    1: 1,
    2: 1
}
def fibonacci(n):
    global counter
    counter += 1
    if n in dictionary:
        return dictionary[n]
    else:
        output = fibonacci(n-1) + fibonacci(n-2)
        dictionary[n] = output
        return output

print(fibonacci(15))
print(dictionary)
print(counter)
```

```python
#p.53 평탄화 연습문제 어렵네
def flatten(data):
    output = []
    for i in data:
        if type(i) == list:
            #output에 flatten값을 더한다?
            output += flatten(i)
        else:
            output.append(i)
    return output

example = [[1,2,3],[4,[5,6]],7,[8,9]]
print(f"원본 : {example}")
print(f"변환 : {flatten(example)}")
            
```

```python
def string_reverse(data , output = None):
    if output is None:
        output = []
    if type(data) == str:
        data = list(data)
    if not data:
        return ''.join(output)
    output.append(data.pop())
    return string_reverse(data, output)

print(string_reverse("hello_world"))
```

```python
#문자열 재귀
def reverse(data):
    if not data:
        return ""
    return data[-1] + reverse(data[:-1])

print(reverse("hello_world"))
```

```python
#재귀로 숫자 더하기
def list_sum (data):
    if len(data) == 1 :
        return data[0]
    return data[0] + list_sum(data[1:])

print(list_sum([1,2,3,4,5]))
```

```python
def get_combinations(arr, n):
    result = []
    
    # base case 1: n개를 다 뽑았을 때
    if n == 0:
        return [[]]
    # base case 2: 더 이상 뽑을 요소가 없을 때
    if not arr:
        return []
    
    # 첫 번째 요소를 선택하는 경우
    for rest in get_combinations(arr[1:], n - 1):
        result.append([arr[0]] + rest)
        
    # 첫 번째 요소를 선택하지 않는 경우
    result.extend(get_combinations(arr[1:], n))
    
    return result

nums = [1, 2, 3, 4]
print(f"4개 중 2개 뽑기: {get_combinations(nums, 2)}")
```

```python
# map(함수, 리스트) -> 해당 함수를 계산해서 반환?
# filter(함수, 리스트) -> 함수에 맞는 값들을 반환?
def power(item):
    return item * item
def under_3(item):
    return item < 3

list_input_a = [1,2,3,4,5]

output_a = map(power, list_input_a)
print(list(map(power, list_input_a)))
print()

output_b = filter(under_3, list_input_a)
print(list(output_b))
```
