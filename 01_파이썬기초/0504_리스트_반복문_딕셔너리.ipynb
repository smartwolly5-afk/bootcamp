```python
#변수 입력
info = input("이름과 근무시간을 입력하시오:")
#이름, 시간 추출
name, work_t = info.split(" ")
work_t=int(work_t)
#출력
if(work_t <= 8):
    print("{}님의 급여는 {}원 입니다.".format(name, work_t * 10000))
else:
    print("{}님의 급여는 {}원 입니다.".format(name, ((8 * 10000) + ((work_t - 8) * 15000))))
```

```python
#변수 입력
stu = input("이름과 입실시간을 입력하시오:")
#이름, 시간 추출
name, time = stu.split(" ")
#시와 분 추출
h , m = time.split(":")
h = int(h)
m = int(m)
#9시 정각까지는 정시출석 그 이후는 지각
if((h >= 9) and (m > 0)):
    print("{}님은 지각입니다.".format(name))
else:
    print("{}님은 정시 출석했습니다.".format(name))
```

```python
import datetime
now = datetime.datetime.now()

#요일 저장
d = ['월','화','수','목','금','토','일']
#변수 입력
stu = input("이름과 학습주제를 입력하시오:")
#이름이랑 주제 나누기
name, subject = stu.split(" ")
#출력
print("{}_{}_{}요일_{}공부.md".format(name,now.date(),d[now.weekday()],subject))
```

```python
name = list()
v_name = list()
new = 0
vip = ["김신엽","홍길동","세종대왕"]
while(1):
    new = input("이름을 입력하세요")
    if(new in vip):
        print("VIP입니다.")
        v_name.append(new)
        print("{} {}".format(v_name, name))
    elif(new in name):
        name.remove(new)
        print("{}님 예약이 취소되었습니다.".format(new))
        print(name)
    elif(new == "1"):
        break
    elif(new == "입장"):
        print("{}입장하세요.".format(name.pop(0)))
    else:
        name.append(new)
        print("{}번째 차례입니다.".format(len(v_name) + len(name)))
        print(name)
```

```python
# 딕셔너리 기본 선언
inventory ={
    "메로나" : [300, 20],
    "비비빅" : [400, 3],
    "죠스바" : [250, 100]
}
#메로나 가격과 재고
print("{}원".format(inventory["메로나"][0]))
print("{}개".format(inventory["메로나"][1]))
print()
#월드콘 키값 및 밸류 추가
inventory["월드콘"] = [500, 7]
#추가 되었는지 확인
for key in inventory:
    print("{}:{}".format(key,inventory[key]))
print()
#keys함수로 키값만 불러와서 list 사용
keys_1= inventory.keys()
print("키 목록 리스트 : {}".format(keys_1))
#list함수 활용하여 키값만 추출해서 list로 만들기
keyl = list(inventory.keys())
print("키 목록 리스트 : {}".format(keyl))
print()
print("밸류값 리스트: {}".format(list(inventory.values())))
```

```python
character = {
    "name": "기사",
    "level": 12,
    "items": {
        "sword": "불꽃의 검",
        "armor": "풀플레이트"
    },
    "skill": ["베기", "세게 베기", "아주 세게 베기"]
}
keyl = list(character)
for key in character:
    #딕셔너리 안에 딕셔너리
    if (type(character[key]) == dict):
        for i in character[key]:
            print("{}: {}".format(i,character[key][i]))
    #딕셔너리 안에 리스트
    elif (type(character[key]) == list):
        for i in character[key]:
            print("{}: {}".format(key,i))
    #딕셔너리 안에 키
    else:
        print("{}: {}".format(key,character[key]))
```
