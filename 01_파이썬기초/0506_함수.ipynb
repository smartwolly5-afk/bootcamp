```python
# 딕셔너리 기본 선언
inventory ={
    "메로나" : [300, 20],
    "비비빅" : [400, 3],
    "죠스바" : [250, 100]
}
#메로나 가격과 재고
print("{}원".format(inventory["메로나"][0]))
print("{}개".format(inventory["메로나"][1]))
print()
#월드콘 키값 및 밸류 추가
inventory["월드콘"] = [500, 7]
#추가 되었는지 확인
for key in inventory:
    print("{}:{}".format(key,inventory[key]))
print()
#keys함수로 키값만 불러와서 list 사용
keys_1= inventory.keys()
print("키 목록 리스트 : {}".format(keys_1))
#list함수 활용하여 키값만 추출해서 list로 만들기
keyl = list(inventory.keys())
print("키 목록 리스트 : {}".format(keyl))
print()
print("밸류값 리스트: {}".format(list(inventory.values())))
```

```python
character = {
    "name": "기사",
    "level": 12,
    "items": {
        "sword": "불꽃의 검",
        "armor": "풀플레이트"
    },
    "skill": ["베기", "세게 베기", "아주 세게 베기"]
}
keyl = list(character)
for key in character:
    #딕셔너리 안에 딕셔너리
    if (type(character[key]) == dict):
        for i in character[key]:
            print("{} : {}".format(i,character[key][i]))
    #딕셔너리 안에 리스트
    elif (type(character[key]) == list):
        for i in character[key]:
            print("{} : {}".format(key,i))
    #딕셔너리 안에 키
    else:
        print("{} : {}".format(key,character[key]))
```

```python
#enumerate() 함수
example_list = ["요소A", "요소B", "요소C"]

print("# 단순 출력")
print(example_list)
print()

print("#enumerate() 함수적용출력")
print(enumerate(example_list))

print("#list() 함수로 강제 변환 출력")
print(list(enumerate(example_list)))
print()

print("반복문과 조합하기")
for i, value in enumerate(example_list):
    print(i, value)
```
