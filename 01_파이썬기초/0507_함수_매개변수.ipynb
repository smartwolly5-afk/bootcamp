```python
#26 ##set 함수나 sort함수 사용하면 2줄임 ㅋㅋ
nums = [5,3,1,2,3,5,4,2,1]
nums_1 = list()
temp = 0
a = 0

for i in nums :
    if(i not in nums_1):
        nums_1.append(i)
print (nums_1)
for j in range(len(nums_1)):
    for i in range(j+1, len(nums_1)):
        if (nums_1[j] > nums_1[i]):
            temp = nums_1[j]
            nums_1[j] = nums_1[i]
            nums_1[i] = temp
print(nums_1)
```

```python
cart = [{"name": "물", "price": 1000, "stock": 5},

{"name": "커피", "price": 3000, "stock": 2},

{"name": "도시락", "price": 5000, "stock": 3}]

want = [{"name": "물", "qty": 6}, {"name": "커피", "qty": 1}, {"name": "도시락", "qty": 2}]
a = 0
price = 0
c = [[] for _ in range(3)] 
w = [[] for _ in range(3)]
d = list()
sum_c = 0
sum_p =0

for a in range(3):
    for i , k in cart[a].items() :
        c[a].append(k)

for a in range(3):
    for i , k in want[a].items() :
        w[a].append(k)
for a in range(3):
    d.append(w[a][1])
print(c,w)
for a in range(3):
    if(c[a][2]<w[a][1]):
        print(f"재고 부족으로 {c[a][0]}은 {c[a][2]}개만 담습니다")
        d[a] = c[a][2]
for a in range(3):
    print(f"{c[a][0]} {d[a]}개: {d[a]*c[a][1]}원")
    sum_p += d[a]*c[a][1]

print("----------------------")
for a in range(3):
    sum_c += d[a]
if (sum_c > 5):
    print(f"할인 전 금액: {sum_p}")
    print(f"할인 적용(10%: -{int(sum_p * 0.1)}")
    print(f"최종 결제 금액: {int(sum_p * 0.9)}")
else:
    print(f"최종 결제 금액: {int(sum_p)}")
```

```python
#가변 매개변수는 무조건 하나만 사용 가능
#어디서부터 어디까지가 가변 매개변수인지 구별이 불가능하기 때문에
#가변 매개변수는 맨 마지막에 쓸것 , 값을 아예 안넣어줘도 됨
def print_n_times(n, *values):
    for i in range(n):
        for value in values:
            print(value)
        print()

print_n_times(3, "안녕하세요", "즐거운", "파이썬 프로그래밍")
```

```python
#가변 변수가 기본 매개변수보다 앞에와야한다
#앞에 가변변수가 온 경우에 기본 변수 선언할때 n=을 붙여야됨
def print_n_times(*values, n=2):
    for i in range(n):
        for value in values:
            print(value)
        print()
        
print_n_times("안녕하세요","즐거운","파이썬 프로그래밍", n=3)
```
