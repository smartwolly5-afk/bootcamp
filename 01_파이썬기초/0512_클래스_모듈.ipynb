```python
#ch07 확인문제 5번 리스트로 구현
items = {"사과":1200, "바나나":800, "딸기":2500}
n = []
c = []
d = []
price = 0
while(1):
    name = input("상품명(종료:q):")
    if name == "q":
        break
    try:
        if name not in items.keys():
            print("-> 오류: 목록에 없는 상품입니다.")
            continue
    except:
        print("무언가 틀렸으니 다시 입력하세요")
        continue
    try:
        count = int(input("수량:"))
    except:
        print("정수를 입력하세요")
        continue
    if count < 1:
        print("1개 미만으로 구매할 수 없습니다.")
        continue
    n.append(name)
    c.append(count)
    name = ""
    count = ""
for i in range(len(n)):
    price += items[n[i]] * c[i]
    d.append(str(n[i])+":"+str(c[i])+"개")
for i in range(len(d)):
    print(f"{d[i]} ",end ="")
print(f"\n총액은: {price}원")
```

```python
class Book:
    def __init__(self,name,author,price):
        self.name = name
        self.author = author
        self.price = price

    def my_info(self):
        print(f"{self.name} / {self.author} / {self.price}원")

    def get_discount(self):
        return self.price * 0.9
    
    def get_high(self):
        if self.price >= 30000:
            return "TRUE"
        else:
            return "False"
        
    def print_all(self):
        self.my_info()
        print(f"할인가: {self.get_discount()}")
        print(f"고가책 여부:{self.get_high()}")


books = [
    Book("김신엽의 인생","김신엽",1000000),
    Book("파이썬 기초","윤영주",30000),
    Book("AI 입문","홍길동",45000),
    Book("데이터 분석","김데이터",28000),
    Book("머신러닝 실전","이인공지능",52000),
    Book("웹개발 스타트","박코딩",22000)
]
for i in books:
    i.print_all()
    print('----------------------------')
```
