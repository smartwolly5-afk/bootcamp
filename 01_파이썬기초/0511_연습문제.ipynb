```python
#csv 파일 형식으로 랜덤 몸무게 저장
import random
hanguls = list("가나다라마바사아자차카타파하")
with open("info.txt","w") as file:
    for i in range(1000):
        name = random.choice(hanguls) + random.choice(hanguls)
        weight = random.randrange(40, 100)
        height = random.randrange(140, 200)
        file.write(f"{name}, {weight}, {height}\n")
```

```python
#p.80 비만 계산
new_line = []
with open("info.txt", "r") as file:
    for line in file :
        (name, weight, height) = line.strip().split(",")
        if(not name) or (not weight) or (not height):
            continue
        bmi = int(weight) /(((int(height)) / 100) ** 2)
        if bmi >= 25:
            new_line.append(f"{name},{weight},{height}, {int(bmi)}, 과체중")
            print(f"이름:{name}\n몸무게:{weight}\n키:{height}\nbmi:{int(bmi)}\n결과:과체중\n")
        elif bmi >= 18.5:
            new_line.append(f"{name},{weight},{height}, {int(bmi)}, 정상")
            print(f"이름:{name}\n몸무게:{weight}\n키:{height}\nbmi:{int(bmi)}\n결과:정상\n")
        else:
            new_line.append(f"{name},{weight},{height}, {int(bmi)}, 저체중")
            print(f"이름:{name}\n몸무게:{weight}\n키:{height}\nbmi:{int(bmi)}\n결과:저체중\n")
with open("new_info.txt", "w") as file:
    for line in new_line:
        file.write(f"{line}\n")

        
```

```python
#4개 더하기
def moving_average_4(a):
    b = 0
    sum_4 = 0
    c = []
    for j in range(len(a)-3):
        for i in range(0+b,4+b):
            sum_4 += a[i]
        c.append(sum_4/4)
        sum_4 = 0 
        b += 1
    return c

my_list =[100, 200, 300, 200, 400, 500, 600, 700]
print(moving_average_4(my_list))
```

```python
#로또 번호 만들기
def make_lotto_without_replacement():
    import random as rd
    a = []
    count = 0
    while(count != 6):
        b= rd.randint(1,45)
        if b not in a:
            a.append(b)
            count += 1
        else:
            continue
    return a

print(make_lotto_without_replacement())
```

```python
def test():
    print("test() 함수의 첫 줄입니다.")
    try:
        print("try 구문이 실행되었습니다.")
        return
        print("try구문의 return 키워드 뒤입니다.")
    except:
        print("except 구문이 실행되었습니다.")
    else:
        print("else 구문이 실행되었습니다.")
    finally:
        print("finally 구문이 실행되었습니다,")
    print("test() 함수의 마지막 줄입니다.")
test()
```
