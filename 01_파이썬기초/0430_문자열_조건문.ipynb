```python
import random as rd
a=input("찾고 싶은 숫자 입력:")
b=str(rd.randint(10^15,99999999999999))
print(b)
start_index=0
indices = []
i=0
while True:
    pos = b.find(a, start_index)

    if pos == -1:
        break
    indices.append(pos)

    start_index = pos + 1

if not indices:
    print("그 숫자가 없습니다.")

else:
    for i in range (len(indices)):
        print("{} 위치의 {}".format(indices[i],b[indices[i]:indices[i]+len(a)]))
```

```python
import random as rd
'''일단 6자리 숫자 받기'''

num=list()
g=0
i=0
while(i!=6):
    b=rd.randint(1,35)
    i += 1
    if(b in num):
        i -= 1
    else:
        num.append(b)

print(num)

a=input("너의 숫자 6개를 중복 없이 입력하시오(예:1 2 3 4 5 6)")
my_num=a.split(" ")
for i in range(6):
    if(int(my_num[i]) in num):
        g += 1
    print(my_num)
if(g>=2):
    print("{}원".format(100**(g-1)))
else:
    print("꽝")
```

```python
#모듈 호출
import datetime
now = datetime.datetime.now()

#요일 추출용 리스트
days = ['월', '화', '수', '목', '금', '토', '일']
#변수 입력
my = input("이름과 학습 주제를 입력해주세요:")
#이름, 주제 추출
[name, subject] = my.split()
#요일 추출
date = str(now.year) + "-" + str(now.month) + "-" + str(now.day)
#출력
print("{}_{}_{}요일_{}공부.md".format(name,date,days[now.weekday()],subject))
```
