```python
import requests
import pandas as pd

# =========================
# API 요청
# =========================

url = "https://opensky-network.org/api/states/all"

response = requests.get(url)

data = response.json()
states = data["states"]
# =========================
# 항공기 데이터 추출
# =========================
columns = [
    "icao24",
    "callsign",
    "origin_country",
    "time_position",
    "last_contact",
    "longitude",
    "latitude",
    "baro_altitude",
    "on_ground",
    "velocity",
    "true_track",
    "vertical_rate",
    "sensors",
    "geo_altitude",
    "squawk",
    "spi",
    "position_source"]
df = pd.DataFrame(states, columns=columns)
# =========================
# 원하는 컬럼만 선택
# =========================
result = df[
    [
        "callsign",
        "origin_country",
        "longitude",
        "latitude",
        "velocity",
        "baro_altitude"
    ]
]
# =========================
# 출력
# =========================

print(result.head(6444))


# =========================
# CSV 저장
# =========================

result.to_csv(
    "aircraft_data.csv",
    index=False,
    encoding="utf-8-sig"
)

print("\nCSV 저장 완료")
```
