```python
wd = webdriver.Chrome()
wd.get("https://pokemonkorea.co.kr/pokedex")
```

```python
html = wd.page_source
souppok = BeautifulSoup(html,'html.parser')
```

```python
#1.현재 화면의 높이 가져오기
last_height = wd.execute_script("return document.body.scrollHeight")

while True:
    # 스크롤을 맨 아래로 내리기
    wd.execute_script("window.scrollTo(0, document.body.scrollHeight);")
    
    time.sleep(1) 
    
    # 스크롤 내린 후의 새로운 높이 계산
    new_height = wd.execute_script("return document.body.scrollHeight")
    
    # 더 이상 내려갈 곳이 없으면(높이가 같으면) 반복 종료
    if new_height == last_height:
        break
    last_height = new_height

# 스크롤이 다 끝난 후 최종 HTML 가져오기
html = wd.page_source
```

```python
souppok = BeautifulSoup(html,'html.parser')
```

```python
# pok_name = souppok.select("#pokedexlist")
# pok_name_all = pok_name[0].text
# print(pok_name_all)
items = souppok.select("#pokedexlist > li")
data_list = []
data_type = defaultdict(list)
for item in items:
    text = item.select_one("h3").text.strip()
    part = text.split()
    num = part[0]
    name = part[1]
    type_get = item.select_one(".badge")
    type_info = type_get.text.strip()
    data_type[type_info].append(name)
    pokemon_info = {
        'num' : num,
        'name': name,
        'type': type_info,
        'img' : item.select_one("img")['src']
    }
    data_list.append(pokemon_info)
pok_tbl = pd.DataFrame(data_list)
pok_tbl.to_csv('포켓몬 도감.csv', encoding='cp949', mode='w', index=True)
```

```python
# 데이터를 리스트에 다 담은 후 데이터프레임으로 변환
pok_tbl = pd.DataFrame(data_list)

# HTML 파일로 저장 (이미지 포함)
html_str = '''
<html>
<head><meta charset="utf-8"><title>포켓몬 도감</title></head>
<body>
<table border="1">
    <tr><th>번호</th><th>이름</th><th>속성</th><th>이미지</th></tr>
'''

for index, row in pok_tbl.iterrows():
    html_str += f"<tr><td>{row['num']}</td><td>{row['name']}</td><td>{row['type']}</td><td><img src='{row['img']}' width='50'></td></tr>"

html_str += "</table></body></html>"

with open('pokedex.html', 'w', encoding='utf-8') as f:
    f.write(html_str)
```
