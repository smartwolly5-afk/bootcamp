```python
data = {
    "국어": [80,90,70,30],
    "영어": [90,70,60,40],
    "수학": [90,60,80,70],
}
columns = ["국어","영어","수학"]
index = ["춘향","몽룡", "향단", "방자"]
df = pd.DataFrame(data, index=index, columns=columns)
```

```python
#1번 수학 점수 시리즈
df["수학"]
```

```python
#2번 국어 영어 데이터프레임
df[["국어","영어"]]
```

```python
#3번 평균점수 추가
df["3과목 평균 점수"] = ((df["수학"]+df["국어"]+df["영어"]) / 3).round(3)
df
```

```python
#4번 방자의 영어점수 80 평균점수 다시
df["영어"]["방자"] = 80
df["3과목 평균 점수"] = ((df["수학"]+df["국어"]+df["영어"]) / 3).round(3)
df
```

```python
#5번 춘향의 점수를 데이터프레임
print(df[:"춘향"])
df.T[["춘향"]]
```
